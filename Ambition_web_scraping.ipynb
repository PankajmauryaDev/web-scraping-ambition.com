```python
# similarly we can do for all

name = []
rating = []
reviews = []
ctype = [] 
salaries = []
interviews = []
jobs = []
benefits = []
Photos = []

for i in company:

    name.append(i.find("h2").text.strip())
    rating.append(i.find("div" , class_ = "rating_star_container").text.strip())
    
    reviews.append(i.find("span" , class_ = "companyCardWrapper__companyRatingCount").text.strip().replace(
        "(","").replace(")" , ""))
    ctype.append(i.find("span" , class_="companyCardWrapper__interLinking").text.strip())

    salaries.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[1].text.strip())
     
    interviews.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[2].text.strip())
    
    jobs.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[3].text.strip())

    benefits.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[4].text.strip())
    
    Photos.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[5].text.strip())

```

```python
 # now we will find the 10 page data. 20x10 = 200 data. 
final = pd.DataFrame()

for j in range(1,11):
    
    url = f"https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={j}"

    headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'}

    rsp = requests.get(url, headers=headers)

    soup = BeautifulSoup(rsp.text, "lxml")

    company = soup.find_all("div" , class_ = "companyCardWrapper")

    name = []
    rating = []
    reviews = []
    ctype = [] 
    salaries = []
    interviews = []
    jobs = []
    benefits = []
    Photos = []
    
    for i in company:
    
        name.append(i.find("h2").text.strip())
        rating.append(i.find("div" , class_ = "rating_star_container").text.strip())
        
        reviews.append(i.find("span" , class_ = "companyCardWrapper__companyRatingCount").text.strip().replace(
            "(","").replace(")" , ""))
        ctype.append(i.find("span" , class_="companyCardWrapper__interLinking").text.strip())
    
        salaries.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[1].text.strip())
         
        interviews.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[2].text.strip())
        
        jobs.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[3].text.strip())
    
        benefits.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[4].text.strip())
        
        Photos.append(i.find_all("span" , class_="companyCardWrapper__ActionCount")[5].text.strip())

    d = {"Name" : name , "rating" : rating , "reviews" : reviews , "interviews" : interviews , "jobs" :jobs,
     "benefits":benefits , "Photos" : Photos} 

    df = pd.DataFrame(d)

    # final = final.append(df)
    final = pd.concat([final, df], ignore_index=True)
    
    final.to_csv("AmbitionBox_Companies.csv")
```
